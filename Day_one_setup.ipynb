```python
import os
from google.colab import userdata
from openai import OpenAI

# Safely fetch the credential from Colab's Secrets Manager
os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")

# Optional: If you belong to a specific Org or Project, retrieve them from secrets
# (Or replace with direct string values like "org-xxxxxxxx" or "proj_xxxxxxx")
org_id = None
try:
    org_id = userdata.get("OPENAI_ORG_ID")
except Exception:
    pass

project_id = None
try:
    project_id = userdata.get("OPENAI_PROJECT_ID")
except Exception:
    pass

# Initialize the official SDK client
client = OpenAI(
    organization=org_id,
    project=project_id
)

# Test the connection
try:
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": "You are a helpful teaching assistant."
            },
            {
                "role": "user",
                "content": "Respond with exactly the word: 'Success!'"
            }
        ],
        temperature=0.0
    )

    print(
        f"\nSetup Complete! Model response: "
        f"{response.choices[0].message.content}"
    )

except Exception as e:
    print(f"\nSetup Failed. Error details: {e}")
    print("\nTip: If you belong to multiple organizations or projects, please create the 'OPENAI_ORG_ID' or 'OPENAI_PROJECT_ID' keys in Colab's Secrets manager and try again.")
```

### 🔑 How to Fix OpenAI Restricted Key Permissions

1. **Go to the Keys Page**
   Visit [platform.openai.com/api-keys](https://platform.openai.com/api-keys).

2. **Identify Your Key Type**
   If your key has permissions explicitly set (a Restricted Key), it must have permission to use models:
   * **Model Capabilities / `model.request`**: Must be enabled.
   * **Model Permissions**: Set to **Write** or **All**.

3. **Quickest Solution**
   If you are the owner of the workspace, create a new **"User key"** (instead of a restricted API key). User keys automatically inherit your full account permissions to call models like `gpt-4o-mini`.
